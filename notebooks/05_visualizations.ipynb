# Task 5 — Create Visualizations (Line, Bar, Scatter)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/loan_data.csv")
df.dropna(inplace=True)
df = df.drop_duplicates().reset_index(drop=True)
```

### Line Chart — Income trend across rows (sorted)

```python
sorted_income = df['income'].sort_values().reset_index(drop=True)

plt.figure(figsize=(9, 4))
plt.plot(sorted_income, color='steelblue')
plt.title("Income Values (Sorted)")
plt.xlabel("Index")
plt.ylabel("Income")
plt.tight_layout()
plt.savefig("line_chart.png")
plt.show()
print("Line chart saved.")
```

### Bar Chart - Loan approval rate by education

```python
approval = df.groupby('education')['approved'].mean() * 100

plt.figure(figsize=(7, 4))
plt.bar(approval.index, approval.values, color=['steelblue','tomato','green'], edgecolor='black')
plt.title("Loan Approval Rate by Education (%)")
plt.xlabel("Education")
plt.ylabel("Approval Rate (%)")
plt.tight_layout()
plt.savefig("bar_chart.png")
plt.show()
print("Bar chart saved.")
```

### Scatter Plot - Income vs Loan Amount

```python
colors = df['approved'].map({1: 'steelblue', 0: 'tomato'})

plt.figure(figsize=(8, 5))
plt.scatter(df['income'], df['loan_amount'], c=colors, alpha=0.5, s=30)
plt.title("Income vs Loan Amount (Blue=Approved, Red=Rejected)")
plt.xlabel("Income")
plt.ylabel("Loan Amount")
plt.tight_layout()
plt.savefig("scatter_plot.png")
plt.show()
print("Scatter plot saved.")
```
