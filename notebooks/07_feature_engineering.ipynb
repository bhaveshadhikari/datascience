# Task 7 — Feature Engineering

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/loan_data.csv")
df.dropna(inplace=True)
df = df.drop_duplicates().reset_index(drop=True)
print("Starting shape:", df.shape)
```

### New Feature 1 — loan_to_income ratio

```python
df['loan_to_income'] = df['loan_amount'] / df['income']
print(df['loan_to_income'].describe().round(3))
```

### New Feature 2 — age group (binning)

```python
df['age_group'] = pd.cut(df['age'], bins=[0, 30, 40, 60],
                          labels=['young', 'mid', 'senior'])
print(df['age_group'].value_counts())
```

### New Feature 3 — high credit score flag

```python
df['good_credit'] = (df['credit_score'] >= 650).astype(int)
print("Good credit:", df['good_credit'].value_counts().to_dict())
```

### Experinece level categorization

```python
df['experience_level'] = pd.cut(
    df['employment_years'],
    bins=[0, 2, 5, 10, 50],
    labels=['Entry', 'Junior', 'Mid', 'Senior']
)
print("Experience level:", df['experience_level'].value_counts().to_dict())
```
