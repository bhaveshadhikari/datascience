# Task 2 — Handle Missing Values Using Multiple Strategies

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/loan_data.csv")
print("Missing before:")
print(df.isnull().sum())
```

### Strategy 1 — Fill with Mean (age)

```python
df['age'].fillna(df['age'].mean(), inplace=True)
print("age missing:", df['age'].isnull().sum())
```

### Strategy 2 — Fill with Median (income)
Median is better when data has outliers — it is not affected by extreme values.

```python
df['income'].fillna(df['income'].median(), inplace=True)
print("income missing:", df['income'].isnull().sum())
```

### Strategy 3 — Fill with Mode (credit_score)
Mode is typically used for categorical columns. Here we use it as a demonstration.

```python
df['credit_score'].fillna(df['credit_score'].mode()[0], inplace=True)
print("credit_score missing:", df['credit_score'].isnull().sum())
```

### Verify — no missing values remain

```python
print("Missing after:")
print(df.isnull().sum())
```
